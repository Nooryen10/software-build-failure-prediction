```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = pd.read_csv('/content/drive/MyDrive/software-build-failure-prediction/data/processed/df_clean.csv')
print(df_clean.shape)
df_clean.head()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='target', data=df_clean)
plt.xticks([0,1], ['Passed', 'Failed'])
plt.title('Class Distribution: Passed vs Failed Builds')
plt.ylabel('Number of Builds')
plt.savefig('/content/drive/MyDrive/software-build-failure-prediction/results/figures/class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
df_clean.describe()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12,8))

sns.boxplot(x='target', y='git_diff_src_churn', data=df_clean, ax=axes[0,0])
axes[0,0].set_title('Source Churn vs Build Outcome')
axes[0,0].set_xticklabels(['Passed','Failed'])

sns.boxplot(x='target', y='gh_diff_files_modified', data=df_clean, ax=axes[0,1])
axes[0,1].set_title('Files Modified vs Build Outcome')
axes[0,1].set_xticklabels(['Passed','Failed'])

sns.boxplot(x='target', y='gh_team_size', data=df_clean, ax=axes[1,0])
axes[1,0].set_title('Team Size vs Build Outcome')
axes[1,0].set_xticklabels(['Passed','Failed'])

sns.boxplot(x='target', y='gh_sloc', data=df_clean, ax=axes[1,1])
axes[1,1].set_title('Repo Size (SLOC) vs Build Outcome')
axes[1,1].set_xticklabels(['Passed','Failed'])

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/software-build-failure-prediction/results/figures/feature_boxplots.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
numeric_df = df_clean.select_dtypes(include='number')
correlations = numeric_df.corr()['target'].sort_values(ascending=False)
print(correlations)
```

```python
plt.figure(figsize=(8,10))
top_corr_features = correlations.abs().sort_values(ascending=False).head(15).index
sns.heatmap(numeric_df[top_corr_features].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap: Top 15 Features vs Target')
plt.savefig('/content/drive/MyDrive/software-build-failure-prediction/results/figures/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
for col in ['git_diff_test_churn', 'gh_diff_tests_added', 'gh_diff_tests_deleted']:
    print(col, df_clean[col].nunique(), df_clean[col].unique()[:5])
```

```python
zero_variance_cols = ['git_diff_test_churn', 'gh_diff_tests_added', 'gh_diff_tests_deleted']
df_clean = df_clean.drop(columns=zero_variance_cols)

print('Updated df_clean shape:', df_clean.shape)

df_clean.to_csv('/content/drive/MyDrive/software-build-failure-prediction/data/processed/df_clean.csv', index=False)
print('df_clean re-saved.')
```

```python
df_leaky = pd.read_csv('/content/drive/MyDrive/software-build-failure-prediction/data/processed/df_leaky.csv')

cols_to_drop = [c for c in zero_variance_cols if c in df_leaky.columns]
df_leaky = df_leaky.drop(columns=cols_to_drop)

print('Updated df_leaky shape:', df_leaky.shape)

df_leaky.to_csv('/content/drive/MyDrive/software-build-failure-prediction/data/processed/df_leaky.csv', index=False)
print('df_leaky re-saved.')
```
